```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('../data/raw/heart_disease_uci.csv')
df = df.drop(columns=['id'])
df = df.drop(columns=['ca', 'slope', 'thal'])
```

```python
df['target'] = (df['num'] > 0).astype(int)
df = df.drop(columns=['num'])
```

```python

for col in ['trestbps', 'chol', 'thalch', 'oldpeak']:
    df[col] = df[col].fillna(df[col].median())

for col in ['fbs', 'restecg', 'exang']:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
cat_cols = ['sex', 'dataset', 'cp', 'fbs', 'restecg', 'exang']
encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le
```

```python

df.to_csv('../data/processed/heart_cleaned.csv', index=False)
print(df.shape)
print(df.isnull().sum().sum(), 'missing values remaining')
```

```python
import joblib
joblib.dump(encoders, '../models/encoders.pkl')
```
